```python
import pandas as pd
import numpy as np
from faker import Faker
import random
from datetime import datetime,timedelta
```

```python
cols_to_drop = [
    'BusinessTravel', 'DailyRate', 'Education', 'EmployeeCount',
    'HourlyRate', 'JobInvolvement', 'MonthlyRate', 'NumCompaniesWorked',
    'Over18', 'OverTime', 'PercentSalaryHike', 'StandardHours',
    'StockOptionLevel', 'TotalWorkingYears', 'TrainingTimesLastYear',
    'YearsInCurrentRole', 'YearsSinceLastPromotion', 'YearsWithCurrManager'
]
```

```python
df = df.drop(columns=cols_to_drop)
```

```python
curr_year = 2026
```

```python
df["HireDate"] = df["YearsAtCompany"].apply(
    lambda years:f"{curr_year-years}-{random.randint(1,12):02d}-{random.randint(1,28):02d}"
)
```

```python
df["HireDate"].head()
```

```python
df['IsActive'] = df['Attrition'].apply(lambda x: False if x == 'Yes' else True)
```

```python
df = df.drop(columns=["Attrition"])
```

```python
dept_data = [
    {'DepartmentID': 1, 'DepartmentName': 'Sales'},
    {'DepartmentID': 2, 'DepartmentName': 'Research & Development'},
    {'DepartmentID': 3, 'DepartmentName': 'Human Resources'}
]
departments_df = pd.DataFrame(dept_data)
```

```python
departments_df.head()
```

```python
jobroles_df = df[['JobRole', 'JobLevel']].drop_duplicates().reset_index(drop=True)
jobroles_df.insert(0, 'JobRoleID', range(1, len(jobroles_df) + 1))
jobroles_df = jobroles_df.rename(columns={'JobRole': 'RoleName'})
```

```python
jobroles_df.head()
```

```python
df = df.merge(departments_df, left_on='Department', right_on='DepartmentName', how='left')
```

```python
df = df.merge(jobroles_df, left_on=['JobRole', 'JobLevel'], right_on=['RoleName', 'JobLevel'], how='left')
```

```python
df.head()
```

```python
df = df.drop(columns=['Department', 'DepartmentName', 'JobRole', 'JobLevel', 'RoleName'])
```

```python
target_rows = 120000
df_scaled = df.sample(n=target_rows, replace=True).reset_index(drop=True)
```

```python
df_scaled["EmployeeID"] = range(1,target_rows+1)
```

```python
fake = Faker()
```

```python
first_names = [fake.first_name() for _ in range(target_rows)]
last_names = [fake.last_name() for _ in range(target_rows)]
emails = [f"{f.lower()}.{l.lower()}{i}@company.com" for i,(f, l) in enumerate(zip(first_names, last_names),)]
```

```python
df_scaled['FirstName'] = first_names
df_scaled['LastName'] = last_names
df_scaled['Email'] = emails
```

```python
employee_cols = [
    'EmployeeID', 'FirstName', 'LastName', 'Email', 'Age', 'Gender',
    'MaritalStatus', 'DistanceFromHome', 'EducationField', 'MonthlyIncome',
    'HireDate', 'IsActive', 'DepartmentID', 'JobRoleID'
]
```

```python
employees_df = df_scaled[employee_cols]
```

```python
employees_df.head()
```

```python
employees_df.to_csv("../data/employees.csv",index=False)
```

```python
review_cols = [
    'EmployeeID', 'PerformanceRating', 'JobSatisfaction',
    'EnvironmentSatisfaction', 'WorkLifeBalance', 'RelationshipSatisfaction'
]
reviews_df = df_scaled[review_cols].copy()
```

```python
reviews_df.insert(0, 'ReviewID', range(1, len(reviews_df) + 1))
```

```python
reviews_df['ReviewDate'] = [
    fake.date_between(start_date='-4y', end_date='today') for _ in range(target_rows)
]
```

```python
reviews_df = reviews_df[[
    'ReviewID', 'EmployeeID', 'ReviewDate', 'PerformanceRating',
    'JobSatisfaction', 'EnvironmentSatisfaction', 'WorkLifeBalance',
    'RelationshipSatisfaction'
]]
```

```python
reviews_df.to_csv('../data/performance_reviews.csv', index=False)
```

```python
num_projects = 1500
projects_data = []
```

```python
for i in range(1, num_projects + 1):
    projects_data.append({
        'ProjectID': i,
        'ProjectName': fake.bs().title(),
        'Budget': random.randint(50000, 1000000),
        'Status': random.choice(['Active', 'Completed', 'Planning'])
    })
```

```python
projects_df = pd.DataFrame(projects_data)
projects_df.to_csv('../data/projects.csv', index=False)
```

```python
num_allocations = 30000

allocations_data = {
    'EmployeeID': [random.randint(1, target_rows) for _ in range(num_allocations)],
    'ProjectID': [random.randint(1, num_projects) for _ in range(num_allocations)],
    'AllocationPercentage': [random.choice([25, 50, 75, 100]) for _ in range(num_allocations)]
}
```

```python
allocations_df = pd.DataFrame(allocations_data)
```

```python
allocations_df = allocations_df.drop_duplicates(subset=['EmployeeID', 'ProjectID'])
```

```python
allocations_df.insert(0,'AllocationID',range(1,len(allocations_df) + 1))
```
