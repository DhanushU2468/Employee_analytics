```python
valid_depts = departments_df['DepartmentID'].tolist()
valid_roles = jobroles_df['JobRoleID'].tolist()
valid_projects = projects_df['ProjectID'].tolist()
```

```python
num_history_emp = len(employees_df)
history_emp_ids = set(employees_df['EmployeeID'])
```

```python
historic_employees = []

for _, row in employees_df.iterrows():
    if row['EmployeeID'] in history_emp_ids:
        # Generate 1 past change per selected employee to keep it simple for the SP
        hist_record = row.to_dict()

        # Reduce salary to simulate a past state (75% to 90% of current)
        hist_record['MonthlyIncome'] = max(1000, int(row['MonthlyIncome'] * random.uniform(0.75, 0.90)))

        # 40% chance they were in a different role historically
        if random.random() < 0.40:
            hist_record['JobRoleID'] = random.choice([r for r in valid_roles if r != row['JobRoleID']])
            # 50% chance the role change also meant a different department
            if random.random() < 0.50:
                hist_record['DepartmentID'] = random.choice([d for d in valid_depts if d != row['DepartmentID']])

        historic_employees.append(hist_record)

historic_employees_df = pd.DataFrame(historic_employees)

# Enforce exact column order matching the Employees table schema
employee_columns = [
    'EmployeeID', 'FirstName', 'LastName', 'Email', 'Age', 'Gender',
    'MaritalStatus', 'DistanceFromHome', 'EducationField', 'MonthlyIncome',
    'HireDate', 'IsActive', 'DepartmentID', 'JobRoleID'
]
historic_employees_df = historic_employees_df[employee_columns]
historic_employees_df.to_csv('../data/historic_employees.csv', index=False)
```

```python
# Create a mapping of valid roles per department based on the existing dataset
# This creates a dictionary like {1: [1, 2], 2: [3, 4, 5], 3: [6, 7]}
dept_role_mapping = employees_df.groupby('DepartmentID')['JobRoleID'].unique().to_dict()
valid_depts = list(dept_role_mapping.keys())

# Select 40% of employees for historical data
num_history_emp = len(employees_df)
history_emp_ids = set(employees_df['EmployeeID'])

historic_employees = []

for _, row in employees_df.iterrows():
    if row['EmployeeID'] in history_emp_ids:
        hist_record = row.to_dict()

        # Reduce salary to simulate a past state (75% to 90% of current)
        hist_record['MonthlyIncome'] = max(1000, int(row['MonthlyIncome'] * random.uniform(0.75, 0.90)))

        # 40% chance they were in a different role/department historically
        if random.random() < 0.40:
            # 50% chance the role change meant a completely different department
            if random.random() < 0.50 and len(valid_depts) > 1:
                new_dept = random.choice([d for d in valid_depts if d != row['DepartmentID']])
                hist_record['DepartmentID'] = new_dept
                # Assign a role that strictly belongs to the newly selected department
                hist_record['JobRoleID'] = random.choice(dept_role_mapping[new_dept])
            else:
                # Same department, different role
                current_dept_roles = dept_role_mapping[row['DepartmentID']]
                available_roles = [r for r in current_dept_roles if r != row['JobRoleID']]

                # If the department has other roles available, pick one; otherwise keep the current role
                if available_roles:
                    hist_record['JobRoleID'] = random.choice(available_roles)

        historic_employees.append(hist_record)

historic_employees_df = pd.DataFrame(historic_employees)

# Enforce exact column order matching the Employees table schema
employee_columns = [
    'EmployeeID', 'FirstName', 'LastName', 'Email', 'Age', 'Gender',
    'MaritalStatus', 'DistanceFromHome', 'EducationField', 'MonthlyIncome',
    'HireDate', 'IsActive', 'DepartmentID', 'JobRoleID'
]
historic_employees_df = historic_employees_df[employee_columns]
historic_employees_df.to_csv('../data/historic_employees.csv', index=False)
```

```python
historic_allocations = []
current_allocs_dict = allocations_df.groupby('EmployeeID')['ProjectID'].apply(set).to_dict()

for emp_id in history_emp_ids:
    num_past_projects = random.randint(1, 3)
    emp_curr_projects = current_allocs_dict.get(emp_id, set())
    available_past_projects = list(set(valid_projects) - emp_curr_projects)

    if available_past_projects:
        past_projects = random.sample(available_past_projects, min(num_past_projects, len(available_past_projects)))

        for proj_id in past_projects:
            historic_allocations.append({
                'EmployeeID': emp_id,
                'ProjectID': proj_id,
                'AllocationPercentage': random.choice([25, 50, 75, 100])
            })

historic_allocations_df = pd.DataFrame(historic_allocations)

# Get the maximum AllocationID from the current allocations dataframe
max_current_allocation_id = allocations_df['AllocationID'].max()

historic_allocations_df = pd.DataFrame(historic_allocations)

# Assign AllocationID starting from max_current_allocation_id + 1
historic_allocations_df.insert(
    0,
    'AllocationID',
    range(max_current_allocation_id + 1, max_current_allocation_id + 1 + len(historic_allocations_df))
)

# Enforce exact column order matching the EmployeeProjects table schema
allocation_columns = ['AllocationID', 'EmployeeID', 'ProjectID', 'AllocationPercentage']
historic_allocations_df = historic_allocations_df[allocation_columns]
historic_allocations_df.to_csv('../data/historic_employee_projects.csv', index=False)
```
